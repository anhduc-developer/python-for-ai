```python
def levenshtein_distance(source, target):
    n = len(source) + 1
    m = len(target) + 1
    distances = [[0] * (m) for _ in range(n)]
    for i in range(1, m):
        distances[0][i] = i
    for i in range(1, n):
        distances[i][0] = i
    for i in range(1, n):
        for j in range(1, m):
            if source[i - 1] == target[j - 1]:
                distances[i][j] = distances[i - 1][j - 1]
            else:
                delcost = distances[i - 1][j] + 1
                inscost = distances[i][j - 1] + 1
                subcost = distances[i - 1][j - 1] + 1
                distances[i][j] = min(delcost, inscost, subcost)
    return distances[-1][-1]
result = levenshtein_distance('hola', 'hello')
print(result)
```

```python
class MyStack:
    def __init__(self, capacity):
        self.__data = []
        self.__capacity = capacity
    def is_full(self):
        return len(self.__data) == self.__capacity
    def is_empty(self):
        return len(self.__data) == 0
    def describe(self):
        print(self.__data)
    def push(self, item):
        if not self.is_full():
            self.__data.append(item)
        else:
            print('Do Nothing!')
    def pop(self):
        if not self.is_empty():
            return self.__data.pop(-1)
        else:
            print('Do Nothing!')
            return None
stack = MyStack(5)
stack.push(4)
stack.push(1)
stack.push(6)
stack.describe()
stack.pop()
stack.pop()
stack.pop()
stack.describe()
stack.pop()
stack.describe()
```
