```python
class Math1:
    def __init__(self, datas):
        self._datas = datas
    def compute_total(self):
        return sum(self._datas)
class Math2(Math1):
    def compute_average(self):
        return self.compute_total() / len(self._datas)
a = Math2([1, 2, 3, 4])
average = a.compute_average()
total = a.compute_total()
print(average, total)
```

```python
class Employee:
    def __init__(self, name, salary):
        self._name = name
        self._salary = salary
    def compute_salary(self):
        return self._salary * 3.0
class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self._bonus = bonus
    def compute_salary(self):
        return super().compute_salary() + self._bonus
manager = Manager('Mai Anh Duc', 1000000, 50000000)
print(manager.compute_salary())
    
```

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def compute_area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
    def say(self):
        print("Hello World")
class Square(Shape):
    def __init__(self, side):
        self.__side = side
    def compute_area(self):
        return self.__side * self.__side
    def perimeter(self):
        return self.__side * 4
class Canva:
    def draw(self, a_shape):
        print(a_shape.compute_area())

#Test
c = Canva()
c.draw(Square(5))

class Rectangle:
    def __init__(self, height, width):
        self.__height = height
        self.__width = width
    def compute_area(self):
        return self.__height * self.__width
r = Rectangle(10, 20)
print(r.compute_area())
c2 = Canva()


class Circle(Shape):
    def __init__(self, radius):
        self.__radius = radius
    def compute_area(self):
        return self.__radius * self.__radius * 3.14
    def perimeter(self):
        return self.__radius * 3.14
c = Circle(10)
print(c.compute_area())
print(c.perimeter())
```

```python
class Father:
    def advise(self):
        print('Work hard!')
class Mother:
    def advise(self):
        print('Take a rest!')
class Child(Mother, Father):
    def __init__(self, name):
        self.__name = name
    def describe(self, condition):
        print(self.__name)
        if condition == True:
            Father.advise(self)
        else:
            Mother.advise(self)
john = Child('John')
john.advise()
```

```python
class Species:
    def describe(self):
        print('I am living!')
class Animal(Species):
    def eat(self):
        print('I am eating')
class Cat(Animal):
    def wander(self):
        print('I am wandering!')
a_cat = Cat()
a_cat.describe()
a_cat.eat()
a_cat.wander()
```

```python
from abc import ABC, abstractmethod
class Bird(ABC):
    @abstractmethod
    def eat(self):
        pass
class FlyingBird(Bird):
    @abstractmethod
    def fly(self):
        pass
class Parrot(FlyingBird):
    def eat(self):
        print('I like eating ...')
    def fly(self):
        print('I can fly...')
class Penguin(Bird):
    def eat(self):
        print('I like eating ...')
parrot = Parrot()
parrot.eat()
parrot.fly()

penguin = Penguin()
penguin.eat()
```
