```python
class Calculator:
    def add(self, a, b):
        self.sum = a + b
        self.a = a
        self.b = b
        return self.sum
    def substract(self, a, b):
        return a - b
my_calculator = Calculator()
print(vars(my_calculator))
print(my_calculator.add(10, 20))
print(vars(my_calculator))
```

```python
class Square:
    def __init__(self, side):
        self.side = side
    def compute_area(self):
        return self.side ** 2
    def compute_perimeter(self):
        return self.side * 2
    def __str__(self):
        return str(self.side)
s1 = Square(2)
s2 = Square(6)
s3 = Square(3)
my_list = [s1, s2, s3]
my_list.sort(key=lambda x : x.compute_area(), reverse=True)
for x in my_list:
    print(x, end = " ")
```

```python
class Calculator:
    def __call__(self, a, b):
        return a + b
calc = Calculator()
print(calc(10, 20))
```

```python
class Animal:
    def __init__(self, name):
        self.name = "Tiger"
    def make_sound(self):
        print("Hzzz: ", self.name)
class Cat(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.name = name
        self.breed = breed
a = Cat("Cao", "Mew")
a.make_sound()
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def make_sound(self):
        return "Some generic animal sound"
class Cat(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
    def info(self):
        return f'{self.name} is Cat of breed {self.breed}'
    def make_sound(self):
        return "Mew"
class Simease(Cat):
    def make_sound(self):
        return "Hello PTIT"
a = Simease("Tiger", "Simeases")
print(a.make_sound())
```

```python
#Multiple Inheritance
class Base1:
    def __init__(self):
        self.str1 = "Base1"
        print("Base1 Initialized")
class Base2:
    def __init__(self):
        self.str2 = "Base2"
        print("Base2 Initialized")
class Derived(Base1, Base2):
    def __init__(self):
        Base1.__init__(self)
        Base2.__init__(self)
        print("Derived Initialized")
    def display(self):
        print(self.str1, self.str2)
obj = Derived()
obj.display()

```

```python
class Math1:
    def is_even(self, number):
        return number % 2 == 0
    def factorial(self, number):
        result = 1
        for i in range(1, number + 1):
            result *= i
        return result
class Math2(Math1):
    def __init__(self):
        super().__init__()
    def estimate_euler(self, number):
        result = 1
        for i in range(1, number + 1):
            result += 1 / self.factorial(i)
        return result
a = Math2()
print(a.estimate_euler(100))
```
