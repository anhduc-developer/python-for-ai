```python
import math
def compute_quadratic_equation(a, b, c):
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        return ()
    elif delta == 0:
        return (-b / (2 * a))
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        return (x1, x2)
a, b, c = map(int, input().split())
print(compute_quadratic_equation(a, b, c))
```
