```python
#1
num_list = [3, 4, 5, 1, -44, 5, 10, 12, 33, 1]
result = []
n, k = len(num_list), 3
for i in range(n - k + 1):
    result.append(max(max(num_list[i:i+3]), 1))
print(result)

#2
ans = []
sub_list = []
for element in num_list:
    sub_list.append(element)
    if len(sub_list) == k:
        ans.append(max(max(sub_list), 1))
        del sub_list[0]
print(ans)
```

```python
from collections import Counter
def Standardlize(s):
    s = s.replace(" ", '')
    return s
with open("data.txt") as file:
    word = file.read()
    file.close()
    word = Standardlize(word)
    result = Counter(word)
    for key, value in result.items():
        print(key, " : ", value)
```

```python
def levenshtein_distance(source, target):
    distances = [[0] * (len(target) + 1) for _ in range(len(source) + 1)]
    for i in range(1, len(source) + 1):
        distances[i][0] = i
    for i in range(1, len(target) + 1):
        distances[0][i] = i
    for i in range(1, len(source) + 1):
        for j in range(1, len(target) + 1):
            if(source[i - 1] == target[j - 1]):
                distances[i][j] = distances[i - 1][j - 1]
            else:
                del_cost = distances[i - 1][j]
                ins_cost = distances[i][j - 1]
                sub_cost = distances[i - 1][j - 1]
                if(del_cost <= ins_cost) and (del_cost <= sub_cost):
                    distances[i][j] = del_cost + 1
                elif(ins_cost <= del_cost) and (ins_cost <= sub_cost):
                    distances[i][j] = ins_cost + 1
                else:
                    distances[i][j] = sub_cost + 1
    edit_distance = distances[len(source)][len(target)]
    print(edit_distance)
levenshtein_distance('hola', 'hello')

```
