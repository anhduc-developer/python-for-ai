```python
def fx(x):
    return x ** 2 - 1
def binary_search_min(fx, a, b, tol=1e-7):
    print("left \t\t right \t\t middle")
    while b > a + tol:
        mid = (a + b) / 2
        print("%.10f \t %.10f \t %.10f" % (a, b, mid))
        left_mid = mid - tol
        right_mid = mid + tol
        if fx(left_mid) < fx(right_mid):
            b = mid
        else:
            a = mid
    return (a + b) / 2
x = binary_search_min(fx, -8, 10)
result = fx(x)
print(result)
result2 = fx(-0.0000000075)
print(result2)
```

```python
def Bubble_sort(data):
    step = 0
    n = len(data)
    for i in range(n):
        for j in range(n - i - 1):
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
                step += 1
    return (step, data)
data = [5, 4, 7, 1, 3 , 2]
step = Bubble_sort(data)[0]
new_data = Bubble_sort(data)[1]
print("step = ", step)
print("new_data = ", new_data)
```
